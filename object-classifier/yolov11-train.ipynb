```python
import torch
from ultralytics import YOLO
# model = YOLO('yolov8x.pt')
# model = YOLO('yolov3.pt')
model = YOLO('yolo11m-cls.pt')
```

```python
model.model.model[0].in_channels = 1
model.train(data='/content/gdrive/MyDrive/tesis/dataset-2025-04-23/dataset.yaml', epochs=200, batch=8, imgsz=640, workers=1)
```
